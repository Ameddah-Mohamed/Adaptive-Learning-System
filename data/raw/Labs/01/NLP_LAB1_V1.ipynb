3. **Visualization**: Finally, we will visualize the top 10 most frequent nouns, verbs, and adjectives using a plot.

```python
import random
from nltk.tokenize import word_tokenize
from nltk import FreqDist, pos_tag
from collections import defaultdict, Counter
import matplotlib.pyplot as plt
from nltk.corpus import gutenberg, stopwords

# Load the "Alice in Wonderland" text from NLTK's Gutenberg corpus
alice_text = gutenberg.raw('carroll-alice.txt')
stop_words = set(stopwords.words('english'))
print("Total characters:", len(alice_text))
print("\nFirst 500 characters:\n", alice_text[:500])


def analyze_book_words():
    """
    Analyzes word frequency and grammatical categories in Alice in Wonderland
    """
    print("\n=== Exercise 1: Book Word Analysis ===")
    words = nltk.word_tokenize(alice_text)
    # removing stop words
    for stop_word in stop_words:
        while stop_word in words:
            words.remove(stop_word)
    # keeping only length>5 for part 1
    part1_words = words
    for word in part1_words:
        if len(word)<5:
            part1_words.remove(word)
    print("1.a counting word occurences using DefaultDict:")
    counts = defaultdict(int)
    for word in part1_words:
        counts[word]+=1
    counter = Counter(counts)
    print(counter.most_common(10))
    print("1.b using FreqDist:")
    freq_dist = nltk.FreqDist(part1_words)
    print(freq_dist.most_common(10))
    
    
    # part 2 : splitting grammatically
    print("\n === part 2 ===")
    tagged_words = nltk.pos_tag(words)
    verbs = [word for word,tag in tagged_words if tag.startswith("VB")]
    nouns = [word for word,tag in tagged_words if tag.startswith("NN")]
    adjs = [word for word,tag in tagged_words if tag.startswith("JJ")]
    frequent_words =  [nltk.FreqDist(verbs).most_common(10),nltk.FreqDist(nouns).most_common(10),nltk.FreqDist(adjs).most_common(10)]
    print(frequent_words)
    
    
    # part 3 : visualization
    print("\n == part 3==")
    # plot all occurences in one bar plot to compare, separate colors for each category
    words = [word for word, count in frequent_words[0]+frequent_words[1]+frequent_words[2]]
    counts = [count for word, count in frequent_words[0]+frequent_words[1]+frequent_words[2]]
    colors = ['r']*10 + ['g']*10 + ['b']*10
    plt.figure(figsize=(12,6))
    plt.bar(words, counts, color=colors)
    plt.xticks(rotation=45)
    plt.title("Top 10 Verbs, Nouns, and Adjectives in Alice in Wonderland")
    plt.xlabel("Words")
    plt.ylabel("Frequency")
    plt.show()
analyze_book_words()
```

```python
import nltk
import random
from nltk.tokenize import word_tokenize
from nltk import FreqDist, pos_tag
from collections import defaultdict, Counter
import matplotlib.pyplot as plt
from nltk.corpus import gutenberg, stopwords
import itertools
# Load the "Alice in Wonderland" text from NLTK's Gutenberg corpus
alice_text = gutenberg.raw('carroll-alice.txt')
stop_words = set(stopwords.words('english'))
def analyze_character_cooccurrence():
    """Analyze character co-occurrence using NLTK sentence tokenization"""
    print("\n=== Exercise 2: Character Co-occurrence Analysis ===")
    paragraphs = alice_text.split('\n\n')
    paragraphs = [nltk.word_tokenize(p) for p in paragraphs if p.strip()]
    main_characters = {'Alice', 'Queen', 'King', 'Rabbit', 'Hatter', 'Duchess'}
    co_occurrence_counts = defaultdict(int)
    for para in paragraphs:
        present_chars = set([word for word in para if word in main_characters])
        for pair in itertools.combinations(sorted(present_chars),2):
            co_occurrence_counts[pair]+=1
            
    print("pair co-occcurences")
    for pair,count in sorted(co_occurrence_counts.items(),key=lambda x: -x[1]):
        print (f"{pair}:{count}")


analyze_character_cooccurrence()
```

```python
import nltk
import random
from nltk.tokenize import word_tokenize
from nltk import FreqDist, pos_tag
from collections import defaultdict, Counter
import matplotlib.pyplot as plt
from nltk.corpus import gutenberg, stopwords

# Load the "Alice in Wonderland" text from NLTK's Gutenberg corpus
alice_text = gutenberg.raw('carroll-alice.txt')
# NOTE: The original code used 'text' later, replacing with 'alice_text' for consistency
text = alice_text
stop_words = set(stopwords.words('english'))

# Tokenize the text and keep all words, including stop words
def preprocess_text(text):
    # Tokenize the text
    tokens = word_tokenize(text) # Added word_tokenize to complete the function
    tokens = [t.lower() for t in tokens]
    return tokens

tokens = preprocess_text(text)

# Function to track the words that come immediately after a given word (Unigram/Bigram model)
def analyze_next_word(tokens):
    next_word_dict = defaultdict(list)
    # Loop through tokens and collect the next words
    for i in range(len(tokens)-1):
        next_word_dict[tokens[i]].append(tokens[i+1])
    # Core: Store the word following token[i] using token[i] as the key
    return next_word_dict

# Function to track the words that come immediately after a pair of words (Trigram model)
def analyze_next_word_trigram(tokens):
    next_word_trigram_dict = defaultdict(list)
    for i in range(len(tokens)-2):
        next_word_trigram_dict[(tokens[i],tokens[i+1])].append(tokens[i+2])
    # Loop through tokens and collect the next word using the pair (token[i], token[i+1]) as the key
    return next_word_trigram_dict

# Function to track the words that come immediately after three words (Four-gram model)
def analyze_next_word_fourgram(tokens):
    next_word_fourgram_dict = defaultdict(list)
    # Loop through tokens and collect the next word using the triplet (token[i], token[i+1], token[i+2]) as the key
    for i in range(len(tokens)-3):
        next_word_fourgram_dict[(tokens[i], tokens[i+1], tokens[i+2])].append(tokens[i+3])
    return next_word_fourgram_dict


# Sentence generation using predicted next words (Bigram, based on 1 previous word)
def generate_sentence_bigram(start_word, next_word_dict, sentence_length=15):
    sentence = [start_word]
    current_word = start_word
    for  _ in range(sentence_length-1):
        followers= next_word_dict.get(current_word,[])
        if not followers:
            break
        most_common = Counter(followers).most_common(5)
        choices = [word for word, _ in most_common]
        next_word = random.choice(choices)
        sentence.append(next_word)
        current_word=next_word
    return ' '.join(sentence)

# Sentence generation using Trigram prediction (based on 2 previous words)
def generate_sentence_trigram(start_pair, next_word_trigram_dict, sentence_length=15):
    sentence = list(start_pair) # start_pair is a tuple/list of two words
    current_pair = start_pair
    # Generate the rest of the sentence by predicting the next word based on the *last two* words
    for _ in range(sentence_length-2):
        followers = next_word_trigram_dict.get(current_pair,[])
        if not followers:
            break
        most_common = Counter(followers).most_common(5)
        choices = [word for word, _ in most_common]
        next_word = random.choice(choices)
        sentence.append(next_word)
        current_pair = (current_pair[1], next_word)
    return ' '.join(sentence[:sentence_length])

# Sentence generation using Four-gram prediction (based on 3 previous words)
def generate_sentence_fourgram(start_triplet, next_word_fourgram_dict, sentence_length=15):
    sentence = list(start_triplet) # start_triplet is a tuple/list of three words
    # Generate the rest of the sentence by predicting the next word based on the *last three* words
    current_triplet = start_triplet
    for i in range(1,sentence_length):
        followers = next_word_fourgram_dict.get(current_triplet,[])
        if not followers:
            break
        most_common = Counter(followers).most_common(5)
        choices = [word for word , _ in most_common]
        next_word = random.choice(choices)
        sentence.append(next_word)
        current_triplet = (current_triplet[1], current_triplet[2], next_word)
    return ' '.join(sentence[:sentence_length])

# Example usage
start_word = 'alice'
start_pair = ('alice', 'was')
start_triplet = ('alice', 'was', 'beginning')


# Step 1: Analyze next word (Bigram/1-word context) - Replaced old analyze_next_word
next_word_dict = analyze_next_word(tokens)

# Step 2: Analyze next word (Trigram/2-word context) - New for the question
next_word_trigram_dict = analyze_next_word_trigram(tokens)

# Step 3: Analyze next word (Four-gram/3-word context) - New for the question
next_word_fourgram_dict = analyze_next_word_fourgram(tokens)


# Step 4: Generate a sentence using Bigram prediction (1-word context)
print("Generated Sentence (Bigram/1-word Context):")
print(generate_sentence_bigram(start_word, next_word_dict))

# Step 5: Generate a sentence using Trigram prediction (2-word context)
print("\nGenerated Sentence (Trigram/2-word Context):")
print(generate_sentence_trigram(start_pair, next_word_trigram_dict))

# Step 6: Generate a sentence using Four-gram prediction (3-word context)
print("\nGenerated Sentence (Four-gram/3-word Context):")
print(generate_sentence_fourgram(start_triplet, next_word_fourgram_dict))
```

```python
import nltk
import random
from nltk.tokenize import word_tokenize
from nltk import FreqDist, pos_tag
from collections import defaultdict, Counter
import matplotlib.pyplot as plt
from nltk.corpus import gutenberg, stopwords
# Load the "Alice in Wonderland" text from NLTK's Gutenberg corpus
alice_text = gutenberg.raw('carroll-alice.txt')
stop_words = set(stopwords.words('english'))

def track_character_actions():
    """Track character actions using POS tagging"""
    tokens = word_tokenize(alice_text)
    tagged_tokens = pos_tag(tokens)
    
    main_characters = {'Alice', 'Queen', 'King', 'Rabbit', 'Hatter', 'Duchess'}
    character_actions = defaultdict(list)
    
    for i, (word, tag) in enumerate(tagged_tokens):
        for character in main_characters:
            if word == character:
                for j in range(i+1, min(i+5, len(tagged_tokens))):
                    if tagged_tokens[j][1].startswith("VB"):
                        character_actions[character].append(tagged_tokens[j][0])
    
    for character in sorted(main_characters):
        if character_actions[character]:
            counts = Counter(character_actions[character])
            print(f"\n{character}:")
            for verb, count in counts.most_common(5):
                print(f"  {verb}: {count}")

track_character_actions()
```

## Exercise 6: Sentence Type Classification

In this exercise, we will categorize sentences in the text based on their type, focusing on **Questions**, **Exclamations**, **Statements**, and **Imperatives**. By analyzing the structure of each sentence, we can gain a better understanding of how characters express themselves and how the narrative develops.

### Tasks

1. **Questions**:
   - Identify sentences that ask for information or clarification.
   - Specifically, we will focus on questions that start with one of the following words:
     - "Who"
     - "What"
     - "Why"
     - "Where"
   - Ensure that these sentences end with a question mark (`?`).

2. **Exclamations**:
   - Identify sentences that express strong emotions or surprise.
   - These sentences typically end with an exclamation mark (`!`).

3. **Statements**:
   - Identify sentences that declare facts or provide information.
   - These sentences usually start with a **noun** or **pronoun** (e.g., "Alice," "She," "The cat").
   - They end with a period (`.`).

4. **Imperatives**:
   - Identify sentences that give commands, requests, or instructions.
   - These sentences usually start with a **verb in its base form** (e.g., "Go," "Run," "Take").
   - They can end with a period (`.`) or an exclamation mark (`!`).


### Tasks to Complete:
- Extract and classify sentences from the text into the four categories mentioned above.
- Print the top 5 sentences from each category to understand the different sentence structures and their usage in the story.


```python
import nltk
import matplotlib.pyplot as plt
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk import pos_tag

# Load the "Alice in Wonderland" text from NLTK's Gutenberg corpus
from nltk.corpus import gutenberg
alice_text = gutenberg.raw('carroll-alice.txt')

# Function to check if a word is a noun or pronoun
def is_noun_or_pronoun(word):
    """Check if a word is a noun or pronoun based on its part of speech tag."""
    tokens = word_tokenize(word)
    tagged = pos_tag(tokens)
    tag = tagged[0][1]
    return tag in ['NN', 'NNP', 'NNPS', 'NNS', 'PRP', 'PRP$']


# Function to extract and categorize sentences
def categorize_sentences(text):
    """Categorize sentences into Questions, Exclamations, Statements, and Imperatives."""
    sentences = sent_tokenize(text)

    # Initialize categories
    categories = {'Questions': [], 'Exclamations': [], 'Statements': [], 'Imperatives': []}

    question_words = {'Who', 'What', 'Why', 'Where'}
    
    for sentence in sentences:
        sentence = sentence.strip()
        if not sentence:
            continue
        
        first_word = sentence.split()[0] if sentence.split() else ""
        tokens = word_tokenize(first_word)
        
        if sentence.endswith('?'):
            if first_word in question_words:
                categories['Questions'].append(sentence)
        elif sentence.endswith('!'):
            categories['Exclamations'].append(sentence)
        elif is_noun_or_pronoun(first_word) and sentence.endswith('.'):
            categories['Statements'].append(sentence)
        elif sentence.endswith('.'):
            tagged = pos_tag([first_word])
            if tagged[0][1].startswith('VB'):
                categories['Imperatives'].append(sentence)
    
    return categories

# Categorize the sentences in the text
categories = categorize_sentences(alice_text)

# Function to plot the frequencies of the categories
def plot_category_frequencies(categories):
    """Plot the frequencies of different sentence categories."""
    cat_names = list(categories.keys())
    frequencies = [len(categories[cat]) for cat in cat_names]
    
    plt.figure(figsize=(10, 6))
    plt.bar(cat_names, frequencies, color=['blue', 'red', 'green', 'orange'])
    plt.xlabel('Sentence Category')
    plt.ylabel('Frequency')
    plt.title('Sentence Types in Text')
    plt.show()

# Print top 5 sentences from each category
for category, sentences in categories.items():
    print(f"{category} ({len(sentences)} total):")
    for i, sentence in enumerate(sentences[:5], 1):
        print(f"  {i}. {sentence[:80]}...")
    print()

# Plot the frequencies
plot_category_frequencies(categories)
```

       - "assault"

2. **Categorize Instances**:
   - After identifying sentences containing the relevant keywords, categorize them into two distinct groups:
     - **Death Indicators**: Sentences containing any of the death-related words.
     - **Violent Actions**: Sentences containing any of the violent action words.

3. **Display Results**:
   - For each category, print the top 10 sentences that correspond to each keyword. This will provide insights into how often these themes appear in the text and in what context.


```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import gutenberg
from collections import defaultdict

# Load Alice in Wonderland text from the NLTK Gutenberg corpus
alice_text = gutenberg.raw('carroll-alice.txt')

# Keywords related to death and violence
death_related_words = {"kill", "murder", "die", "death"}
violent_action_words = {"attack", "stab", "shoot", "injure", "hurt", "blood", "assault"}


def analyze_violence_and_death(text):
    """Identify and categorize sentences with violent actions or death mentions."""
    sentences = sent_tokenize(text)
    
    death_indicators = defaultdict(list)
    violent_actions = defaultdict(list)
    
    for sentence in sentences:
        sentence_lower = sentence.lower()
        tokens = word_tokenize(sentence_lower)
        
        # Check for death-related words
        for word in death_related_words:
            if word in tokens:
                death_indicators[word].append(sentence)
        
        # Check for violent action words
        for word in violent_action_words:
            if word in tokens:
                violent_actions[word].append(sentence)
    
    return death_indicators, violent_actions


# Analyze the text
death_indicators, violent_actions = analyze_violence_and_death(alice_text)


print("DEATH INDICATORS:")
for word in death_related_words:
    if death_indicators[word]:
        print(f"\n'{word}' ({len(death_indicators[word])} occurrences):")
        for i, sentence in enumerate(death_indicators[word][:10], 1):
            print(f"  {i}. {sentence[:80]}...")

print("VIOLENT ACTIONS:")
for word in violent_action_words:
    if violent_actions[word]:
        print(f"\n'{word}' ({len(violent_actions[word])} occurrences):")
        for i, sentence in enumerate(violent_actions[word][:10], 1):
            print(f"  {i}. {sentence}...")
```
