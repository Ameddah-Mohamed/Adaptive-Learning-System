## Generation Parameters
Experiment with these parameters in the `generate()` function:

- **`max_new_tokens`**: Maximum number of tokens to generate
- **`temperature`**: Controls randomness (0.2 = deterministic, >1.0 = creative)
- **`top_p`**: Nucleus sampling threshold (0.9 = consider top 90% probability mass)
- **`top_k`**: Limits sampling to top k tokens (50 = sample from top 50 options)
- **`num_beams`**: Number of beams for beam search (higher = better quality, more compute)

## General Tips

- Compare models systematically using the same prompts
- Keep generation parameters consistent when comparing prompts
- Use clear system roles to set expectations and constraints

````python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
import warnings

# Suppress warnings for cleaner output
warnings.filterwarnings('ignore')

# ============================================================================
# SETUP: Load Models and Tokenizers
# ============================================================================

MODEL_name = "unsloth/gemma-3-4b-it"
    # "unsloth/Phi-3-mini-4k-instruct"
    # "unsloth/Llama-3.2-3B-Instruct"


def load_model_and_tokenizer(model_name):
    """Load a model and its tokenizer with stability fixes."""
    print(f"Loading {model_name}...")

    # Determine the best data type for stability
    # bfloat16 is much more stable than float16 for Llama-3/Gemma models
    dtype = torch.bfloat16 if torch.cuda.is_bf16_supported() else torch.float16

    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModelForCausalLM.from_pretrained(
        model_name,
        torch_dtype=dtype,
        device_map="auto"
    )

    # FIX: Ensure pad token is set (prevents generation errors)
    if tokenizer.pad_token_id is None:
        tokenizer.pad_token_id = tokenizer.eos_token_id
    if model.config.pad_token_id is None:
        model.config.pad_token_id = tokenizer.eos_token_id

    return model, tokenizer

def generate_response(model, tokenizer, messages, **gen_params):
    """Generate a response using the chat template format."""

    # Apply chat template
    # We wrap this in a try/except because some models (like older Gemma versions)
    # might reject 'system' roles in the strict template.
    try:
        formatted_prompt = tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=True
        )
    except Exception as e:
        # Fallback: If system role fails, merge system prompt into first user message
        if messages[0]['role'] == 'system':
            system_content = messages[0]['content']
            # Create a new list without the system message
            new_messages = [{"role": "user", "content": f"System Instruction: {system_content}\n\nUser Query: {messages[1]['content']}"}]
            # Add remaining messages
            new_messages.extend(messages[2:])
            formatted_prompt = tokenizer.apply_chat_template(
                new_messages,
                tokenize=False,
                add_generation_prompt=True
            )
        else:
            raise e

    # Tokenize
    inputs = tokenizer(formatted_prompt, return_tensors="pt").to(model.device)

    # Default generation parameters
    default_params = {
        'max_new_tokens': 200,
        'temperature': 0.7,
        'top_p': 0.9,
        'top_k': 50,
        'num_beams': 1,
        'do_sample': True,
        'pad_token_id': tokenizer.pad_token_id,
        'eos_token_id': tokenizer.eos_token_id
    }
    default_params.update(gen_params)

    # Generate
    with torch.no_grad():
        outputs = model.generate(**inputs, **default_params)

    # Decode only the new tokens
    response = tokenizer.decode(outputs[0][inputs['input_ids'].shape[1]:], skip_special_tokens=True)
    return response.strip()

# Load the first model
model, tokenizer = load_model_and_tokenizer(MODEL_name)


````

````python
test_case = {
    "name": "Analogy-based",
    "messages": [
        {"role": "system", "content": "You are a creative educator who explains concepts using analogies."},
        {"role": "user", "content": "Explain how gradient descent works using an analogy."}
    ]
}


print(f"Test Case: {test_case['name']}")
response = generate_response(model, tokenizer, test_case['messages'])
print(f"\nResponse:\n{response}\n")

# Optional: Temperature experiment on the last case
print("\nTemperature Experiment:")
for temp in [0.2, 0.7, 1.2]:
    response = generate_response(
        model, tokenizer, test_case['messages'],
        temperature=temp, max_new_tokens=500
    )
    print(f"\nTemp={temp}: {response[:500]}...")
````
