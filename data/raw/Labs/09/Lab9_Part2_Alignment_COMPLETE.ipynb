## Lab 8 - Part 2: Alignment with Language Models

### Alignment in Language Models

Alignment is the process of ensuring that language models produce responses consistent with human expectations, ethical standards, and task-specific requirements. It involves using datasets, evaluation metrics, and fine-tuning to adjust the model's behavior.

#### Why Align Language Models?

1. **Relevance**: Aligning models with specific tasks improves their performance and relevance.
2. **Ethics**: Helps ensure models avoid generating harmful or biased content.
3. **Usability**: Tailoring models to user needs enhances their practical utility.

#### The Impact of Alignment: Base vs. Instruct Models

Without alignment, language models struggle to follow instructions and engage in natural conversations. They are trained primarily on text completion tasks, which makes them good at continuing text but poor at understanding and responding to user requests. Alignment training transforms these base models into instruction-following conversational agents.

1. **Base Models (e.g., GPT-2, DistilGPT2, LLaMA)**
   - **Description**: Base models are trained on massive text corpora to predict the next word in a sequence. While powerful at text generation, they are not optimized for following instructions or having conversations.
   - **Behavior**: Without instruction fine-tuning, base models often:
     - Fail to understand what the user is asking for
     - Continue text as if completing a document rather than answering a question
     - Ignore the conversational format entirely
     - Produce irrelevant or off-topic responses
   - **Example**: When asked *"What is the capital of France?"*, a base model might continue with: *"What is the capital of Germany? What is the capital of Italy?"* instead of simply answering "Paris".
   - **Chat Behavior**: Base models cannot engage in proper chat interactions. They don't understand the instruction-response format and will often produce confusing or nonsensical outputs when used as chatbots.

2. **Aligned Models (e.g., ChatGPT, GPT-3.5-Turbo, LLaMA-Instruct)**
   - **Description**: Aligned models are fine-tuned on instruction-following datasets and often further refined using reinforcement learning from human feedback (RLHF). This training teaches them to understand user intent and respond appropriately.
   - **Improvements**: After alignment training, models:
     - Understand and follow explicit instructions
     - Recognize conversational context and respond naturally
     - Provide direct, relevant answers to questions
     - Maintain consistent chat behavior across different topics
   - **Example**: When asked *"What is the capital of France?"*, an aligned model responds directly: *"The capital of France is Paris."*
   - **Chat Behavior**: Aligned models excel at conversational interactions, understanding turn-taking, context, and user intent, making them suitable for chatbot applications.

#### Key Components of Alignment

1. **Datasets**:
   - Curating datasets with diverse, high-quality examples helps align models with human values and task-specific needs.
   - Ethical considerations in dataset selection play a critical role in mitigating biases.

2. **Evaluation Metrics**:
   - Metrics such as accuracy, relevance, and fairness are used to assess how well a model aligns with desired behaviors.
   - Human evaluations often complement automated metrics for nuanced assessment.
   - **Perplexity**: Measures how well a model predicts a sample of text. Lower perplexity indicates better prediction and understanding.

3. **Fine-Tuning Techniques**:
   - Instruction tuning: Using task-specific datasets to teach the model to follow instructions better.
   - Reinforcement Learning from Human Feedback (RLHF): Leveraging feedback loops to iteratively improve model behavior.

### Reinforcement Learning from Human Feedback (RLHF)

Reinforcement Learning from Human Feedback (RLHF) is a technique used to align language models with human expectations by incorporating feedback from human evaluators into the training process. It combines supervised fine-tuning and reinforcement learning to improve the quality, relevance, and ethical considerations of model outputs.

#### Key Steps in RLHF

1. **Supervised Fine-Tuning**
   - The base language model is first fine-tuned on a labeled dataset of high-quality examples.
   - Human annotators provide a set of instructions and corresponding ideal responses.
   - **Goal**: Teach the model to follow instructions and produce useful, contextually appropriate outputs.

2. **Feedback Collection**
   - Human evaluators rank multiple responses generated by the model for the same prompt.
   - These rankings indicate preferences, such as which response is more accurate, ethical, or user-friendly.

3. **Reward Model Training**
   - A separate reward model is trained to predict the human preference rankings.
   - Input: Model responses and prompts.
   - Output: A reward score that reflects how well the response aligns with human preferences.

4. **Reinforcement Learning**
   - The language model is fine-tuned further using a reinforcement learning algorithm, typically Proximal Policy Optimization (PPO).
   - The reward model guides the optimization process, encouraging the model to generate responses that maximize the reward score.

## Implementation

Follow the structure below to complete the lab.

```python
# -------------------------------------------------------------------------
# 1. LIBRARY IMPORTS
# -------------------------------------------------------------------------
# Import all necessary libraries:
# - transformers: AutoTokenizer, AutoModelForCausalLM, Trainer, TrainingArguments, DataCollatorForLanguageModeling
# - datasets: load_dataset
# - torch: for device management
# - math: for calculating perplexity from loss

from transformers import AutoTokenizer, AutoModelForCausalLM, Trainer, TrainingArguments, DataCollatorForLanguageModeling
from datasets import load_dataset
import torch
import math
```

```python
# -------------------------------------------------------------------------
# 2. HYPERPARAMETERS AND CONFIGURATION
# -------------------------------------------------------------------------
# Define all key hyperparameters:
# - MODEL_CHECKPOINT (e.g., "distilgpt2")
# - DATASET_NAME (e.g., "tatsu-lab/alpaca")
# - MAX_LENGTH for tokenization
# - NUM_SAMPLES for dataset size
# - NUM_EPOCHS for training
# - OUTPUT_DIR for saving the fine-tuned model

MODEL_CHECKPOINT = "distilgpt2"
DATASET_NAME = "tatsu-lab/alpaca"
MAX_LENGTH = 256
NUM_SAMPLES = 8000
NUM_EPOCHS = 1
OUTPUT_DIR = "./aligned_distilgpt2"
```

```python
# -------------------------------------------------------------------------
# 3. DEVICE SETUP
# -------------------------------------------------------------------------
# Check for available device (CUDA GPU if available, otherwise CPU) and print:

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"✅ Using device: {device}")
```

```python
# -------------------------------------------------------------------------
# 4. LOAD ALPACA DATASET (New Block)
# -------------------------------------------------------------------------
print("\n--- Loading Instruction Tuning Data ---")

# Load the Alpaca dataset from Hugging Face and select NUM_SAMPLES examples:
dataset = load_dataset(DATASET_NAME, split=f"train[:{NUM_SAMPLES}]")

# -------------------------------------------------------------------------
# 4.1 INSPECT DATASET CONTENT (New Block)
# -------------------------------------------------------------------------

print("\n--- Dataset Content Snapshot ---")

# 1. Print the total number of samples loaded
print(f"Total samples loaded: {len(dataset)}")

# 2. Print the column names (features) of the dataset
print(f"Dataset features: {dataset.column_names}")

# 3. Print the first few examples to inspect the data structure
print("\nFirst 3 examples:")
for i in range(3):
    example = dataset[i]
    print(f"--- Example {i+1} ---")
    print(f"Instruction: {example['instruction']}")
    print(f"Input: {example['input']}")
    print(f"Output: {example['output']}")

# The 'input' field is often empty in this dataset, so you will see a blank line for it in many examples.

```

```python
# -------------------------------------------------------------------------
# 5. DATA PREPARATION (Alpaca Dataset)
# -------------------------------------------------------------------------
# Define a formatting function (format_alpaca) that takes an example (instruction, input, output)
# and formats it into the required text sequence for Causal Language Modeling (CLM),
# e.g., "Instruction: {instruction}\nContext: {input}\nResponse: {output}<|endoftext|>"
# Ensure to handle cases where 'input' (context) is empty.

def format_alpaca(example):
    context = f"\nContext: {example['input']}" if example['input'] else ""
    text = f"Instruction: {example['instruction']}{context}\nResponse: {example['output']}<|endoftext|>"
    return {"text": text}

# Apply the formatting function to the dataset using .map(), removing original columns:
formatted_dataset = dataset.map(format_alpaca, remove_columns=dataset.column_names)

# Split the formatted dataset into training (90%) and evaluation (10%) sets using train_test_split:
split_dataset = formatted_dataset.train_test_split(test_size=0.1, seed=42)
train_dataset = split_dataset["train"]
eval_dataset = split_dataset["test"]

# -------------------------------------------------------------------------
# 6. MODEL AND TOKENIZER LOADING
# -------------------------------------------------------------------------

# Load the AutoTokenizer from the defined MODEL_CHECKPOINT:
tokenizer = AutoTokenizer.from_pretrained(MODEL_CHECKPOINT)

# Set the tokenizer's padding token to be the End-of-Sequence (EOS) token:
tokenizer.pad_token = tokenizer.eos_token

# Load the AutoModelForCausalLM from the MODEL_CHECKPOINT and move it to the detected device:
model = AutoModelForCausalLM.from_pretrained(MODEL_CHECKPOINT).to(device)

# Define a tokenization function (tokenize_function) that processes text examples:
# - Uses the tokenizer to convert text to input_ids.
# - Sets truncation=True and padding="max_length" up to MAX_LENGTH.
def tokenize_function(examples):
    return tokenizer(examples["text"], truncation=True, padding="max_length", max_length=MAX_LENGTH)

# Apply the tokenization function to the training and evaluation datasets using .map(), removing the original 'text' column:
tokenized_train = train_dataset.map(tokenize_function, batched=True, remove_columns=["text"])
tokenized_eval = eval_dataset.map(tokenize_function, batched=True, remove_columns=["text"])

# -------------------------------------------------------------------------
# 7. EVALUATION HELPERS (Functions for Part 1 & Part 2)
# -------------------------------------------------------------------------

# Define the compute_perplexity function:
# - Takes model and tokenized dataset as input.
# - Sets model to evaluation mode (model.eval()).
# - Calculates the average loss over a small subset of the dataset (e.g., first 30 examples).
# - Returns the perplexity (exponent of the average loss).
def compute_perplexity(model, dataset):
    model.eval()
    total_loss = 0
    num_examples = min(30, len(dataset))
    
    with torch.no_grad():
        for i in range(num_examples):
            inputs = {k: torch.tensor(v).unsqueeze(0).to(device) for k, v in dataset[i].items()}
            outputs = model(**inputs, labels=inputs["input_ids"])
            total_loss += outputs.loss.item()
    
    avg_loss = total_loss / num_examples
    return math.exp(avg_loss)

# Define the generate_response function:
# - Takes model and a sample instruction string as input.
# - Sets model to evaluation mode (model.eval()).
# - Formats the instruction into a prompt: "Instruction: {instruction}\nResponse:".
# - Tokenizes the prompt and moves inputs to the device.
# - Uses model.generate() with decoding arguments (e.g., max_new_tokens, temperature, top_p, repetition_penalty).
# - Decodes the output and strips the original prompt to return only the generated response.
def generate_response(model, instruction):
    model.eval()
    prompt = f"Instruction: {instruction}\nResponse:"
    inputs = tokenizer(prompt, return_tensors="pt").to(device)
    
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=100,
            temperature=0.7,
            top_p=0.9,
            repetition_penalty=1.2,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id
        )
    
    response = tokenizer.decode(outputs[0], skip_special_tokens=True)
    return response[len(prompt):].strip()

# Define the list of sample questions for qualitative assessment:
# - Factual (e.g., "What is the capital of France?")
# - Creative (e.g., "Tell me a joke.")
# - Procedural (e.g., "How do I make a sandwich?")
sample_questions = [
    "What is the capital of France?",
    "Tell me a joke.",
    "How do I make a sandwich?"
]

# -------------------------------------------------------------------------
# 8. PART 1: PRE-ALIGNMENT BASELINE EVALUATION
# -------------------------------------------------------------------------

print("\n--- PART 1: Baseline Evaluation ---")

# Calculate and print the Baseline Perplexity using compute_perplexity on the base model:
baseline_perplexity = compute_perplexity(model, tokenized_eval)
print(f"Baseline Perplexity: {baseline_perplexity:.2f}")

# Generate and print the Baseline Answers for the sample questions using generate_response:
# - Store these baseline responses for comparison later.
print("\nBaseline Responses:")
baseline_responses = {}
for question in sample_questions:
    response = generate_response(model, question)
    baseline_responses[question] = response
    print(f"Q: {question}")
    print(f"A: {response}\n")

# -------------------------------------------------------------------------
# 9. PART 2: INSTRUCTION FINE-TUNING
# -------------------------------------------------------------------------

print("\n--- PART 2: Instruction Fine-Tuning ---")

# Define TrainingArguments:
# - output_dir, eval_strategy, save_strategy ("epoch")
# - learning_rate (2e-5)
# - batch_size (per_device_train_batch_size=16)
# - num_train_epochs (NUM_EPOCHS=1)
# - Enable fp16=True for mixed precision.
training_args = TrainingArguments(
    output_dir=OUTPUT_DIR,
    eval_strategy="epoch",
    save_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    num_train_epochs=NUM_EPOCHS,
    fp16=torch.cuda.is_available(),
    logging_steps=100,
    save_total_limit=2
)

# Initialize the Trainer:
# - Pass the model, TrainingArguments, train_dataset, eval_dataset.
# - Use DataCollatorForLanguageModeling (mlm=False) to stack batches.
data_collator = DataCollatorForLanguageModeling(tokenizer=tokenizer, mlm=False)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train,
    eval_dataset=tokenized_eval,
    data_collator=data_collator
)

# Start the training process:
# trainer.train()
trainer.train()

# Save the final fine-tuned model and tokenizer to the OUTPUT_DIR:
model.save_pretrained(OUTPUT_DIR)
tokenizer.save_pretrained(OUTPUT_DIR)
print(f"\n✅ Model saved to {OUTPUT_DIR}")

# -------------------------------------------------------------------------
# 10. POST-ALIGNMENT EVALUATION AND REPORT
# -------------------------------------------------------------------------

print("\n--- Post-Alignment Evaluation ---")

# Calculate and print the Fine-Tuned Perplexity:
finetuned_perplexity = compute_perplexity(model, tokenized_eval)
print(f"Fine-Tuned Perplexity: {finetuned_perplexity:.2f}")

# Generate and print the Fine-Tuned Answers for the same sample questions:
# - Store these post-alignment responses.
print("\nFine-Tuned Responses:")
finetuned_responses = {}
for question in sample_questions:
    response = generate_response(model, question)
    finetuned_responses[question] = response
    print(f"Q: {question}")
    print(f"A: {response}\n")

# Print the Final Comparison Report:
# - Tabulate the Perplexity (Before vs. After).
# - Side-by-side comparison of the generated answers (Before vs. After) for the sample questions.
print("\n" + "="*80)
print("FINAL COMPARISON REPORT")
print("="*80)
print(f"\nPerplexity:")
print(f"  Baseline:    {baseline_perplexity:.2f}")
print(f"  Fine-Tuned:  {finetuned_perplexity:.2f}")
print(f"  Improvement: {baseline_perplexity - finetuned_perplexity:.2f}")
print("\n" + "-"*80)
print("Response Comparison:")
print("-"*80)
for question in sample_questions:
    print(f"\nQuestion: {question}")
    print(f"\n[BASELINE]\n{baseline_responses[question]}")
    print(f"\n[FINE-TUNED]\n{finetuned_responses[question]}")
    print("-"*80)

# -------------------------------------------------------------------------
# 11. INTERACTIVE CHAT INTERFACE
# -------------------------------------------------------------------------

# Define the chat_with_model function:
# - Load the final fine-tuned model and tokenizer from the saved checkpoint.
# - Start an infinite loop to handle user input.
# - Take 'User: ' input. Check for 'exit' or 'quit'.
# - Format the user input into the prompt: "Instruction: {user_input}\nResponse:".
# - Generate the response using chat generation parameters.
# - Decode and clean the output to print the 'Bot:' response.
def chat_with_model():
    print("\n" + "="*80)
    print("INTERACTIVE CHAT INTERFACE")
    print("="*80)
    print("Type 'exit' or 'quit' to end the conversation.\n")
    
    chat_model = AutoModelForCausalLM.from_pretrained(OUTPUT_DIR).to(device)
    chat_tokenizer = AutoTokenizer.from_pretrained(OUTPUT_DIR)
    chat_model.eval()
    
    while True:
        user_input = input("User: ").strip()
        if user_input.lower() in ["exit", "quit"]:
            print("Goodbye!")
            break
        
        prompt = f"Instruction: {user_input}\nResponse:"
        inputs = chat_tokenizer(prompt, return_tensors="pt").to(device)
        
        with torch.no_grad():
            outputs = chat_model.generate(
                **inputs,
                max_new_tokens=150,
                temperature=0.7,
                top_p=0.9,
                repetition_penalty=1.2,
                do_sample=True,
                pad_token_id=chat_tokenizer.eos_token_id
            )
        
        response = chat_tokenizer.decode(outputs[0], skip_special_tokens=True)
        bot_response = response[len(prompt):].strip()
        print(f"Bot: {bot_response}\n")

# Execute the chat_with_model() function to allow interactive testing.
chat_with_model()
```
